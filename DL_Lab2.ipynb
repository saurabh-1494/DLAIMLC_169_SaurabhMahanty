```python
X= [75, 65, 80]
W = [4, 3, 5]
b=-1000
lr=0.01
y=1
```

```python
z=0
for i in range(len(X)):
  z+=X[i]*W[i]
z+=b
print("Net Input(z):",z)
```

```python
y_pred=1 if z>=0 else 0
print("Predicted Output:",y_pred)
```

```python
error = y-y_pred
print("Error:",error)
```

```python
if error!=0:
  for i in range(len(W)):
    W[i]=W[i]+lr*error*X[i]
```

```python
b=b+lr*error
```

```python
z_new=0
for i in range(len(X)):
    z_new+=X[i]*W[i]
z_new+=b
```

```python
y_new=1 if z_new>=0 else 0
```
